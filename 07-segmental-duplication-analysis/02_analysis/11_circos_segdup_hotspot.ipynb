```python
## Get the gff features
gff_feat = Gff(f"{PROJ_ROOT}/code/command-line-script/annotation-merging/output/hifiasm-041425-denovoEnhanced_peaks2utr_sorted.agat.gff3")

seqid2features = gff_feat.get_seqid2features(feature_type=None)
seqid2features

## Get the segdup links 
bedpe=df.copy()
```

```python
## This is filtering the features to find the genes that contains these gene signatures 
matching_features = {}

for seqid, features in seqid2features.items():
    for feature in features:
        if hasattr(feature, 'qualifiers') and feature.qualifiers:
            # Search in the most relevant qualifier keys
            relevant_keys = ['gene', 'product', 'Name', 'ID']
            
            for key in relevant_keys:
                if key in feature.qualifiers:
                    for value in feature.qualifiers[key]:
                        value_str = str(value)
                        # Check if any search term is a substring of the value
                        if any(term in value_str for term in search_terms):
                            if seqid not in matching_features:
                                matching_features[seqid] = []
                            matching_features[seqid].append(feature)
                            break
                    else:
                        continue
                    break  # Break outer loop if match found

print(f"Found features across {len(matching_features)} sequences:")
for seqid, features in matching_features.items():
    print(f"  {seqid}: {len(features)} features")
```

```python
# Extract the gene regions from matching_features
gene_regions = {}
for seqid, features in matching_features.items():
    for feature in features:
        if feature.type == "gene":
            gene_name = None
            # Extract gene name from qualifiers
            if hasattr(feature, 'qualifiers') and feature.qualifiers:
                for key in ['gene', 'Name', 'ID']:
                    if key in feature.qualifiers:
                        gene_name = feature.qualifiers[key][0]
                        break
            
            if gene_name:
                gene_regions[gene_name] = {
                    'chrom': seqid,
                    'start': feature.location.start,
                    'end': feature.location.end,
                    'strand': feature.location.strand
                }

print("Gene regions found:")
for gene, region in gene_regions.items():
    print(f"{gene}: {region['chrom']}:{region['start']}-{region['end']}")
```

```python
# Find segdup links that overlap these gene regions - return as DataFrame
def find_containing_segdups(gene_regions, bedpe_df):
    # Create a list to store matching row indices
    matching_indices = []
    
    for gene_name, region in gene_regions.items():
        chrom = region['chrom']
        start = region['start']
        end = region['end']
        
        # Find segdup links where this region overlaps either region1 or region2
        mask = (
            ((bedpe_df['chr1'] == chrom) & (bedpe_df['start1'] < end) & (bedpe_df['end1'] > start)) |
            ((bedpe_df['chr2'] == chrom) & (bedpe_df['start2'] < end) & (bedpe_df['end2'] > start))
        )
        
        # Get indices of matching rows
        gene_matching_indices = bedpe_df[mask].index.tolist()
        matching_indices.extend(gene_matching_indices)
    
    # Remove duplicates and return the filtered DataFrame
    unique_indices = list(set(matching_indices))
    containing_links_df = bedpe_df.loc[unique_indices].copy()
    
    # Add gene information to the DataFrame
    gene_info = []
    for idx in unique_indices:
        row = bedpe_df.loc[idx]
        genes_in_link = []
        for gene_name, region in gene_regions.items():
            chrom = region['chrom']
            start = region['start']
            end = region['end']
            if ((row['chr1'] == chrom and row['start1'] < end and row['end1'] > start) or
                (row['chr2'] == chrom and row['start2'] < end and row['end2'] > start)):
                genes_in_link.append(gene_name)
        gene_info.append(', '.join(genes_in_link))
    
    containing_links_df['contained_genes'] = gene_info
    
    return containing_links_df

# Get the containing segdup links as DataFrame
containing_links_df = find_containing_segdups(gene_regions, bedpe)
```

```python
size = 12

# Define gene colors
gene_colors = {
    "Cct7": "royalblue",
    "Cct4": "limegreen",
    "Cct3": "purple",
    "Larp7": "firebrick"
}

search_terms = ["Larp7", "Cct4", "Cct7", "Cct3"]


# Suppose karyotype is a DataFrame with 'chr' and 'length' columns
# 1. Read the .fai file
fai = pd.read_csv(
    f"{PROJ_ROOT}/data/denovo_OctDegus_genome/041425-assembly/hifiasm-041425-assembly-mitoFiltered-scaffolded-curated-masked-chrNameAssigned/assembly_final.sorted.headerRenamed.chrAssigned.hardMasked.mito.fasta.fai",
    sep="\t",
    header=None,
    names=["chr", "length", "offset", "linebases", "linewidth"]
)

# 2. Filter for chromosomes and keep first 30
karyotype = fai[
    fai["chr"].str.contains("chr")
].head(30)[["chr", "length"]]

# 3. Set gap degrees
gap_degrees = 5
gap_after = [1] * (len(karyotype) - 1) + [gap_degrees]

chr_sizes = dict(
    zip(
        karyotype["chr"],
        karyotype["length"]
    )
)

circos = Circos(
    sectors=chr_sizes,
    start=45,
    end=359,
    space=2,
    endspace=False
)


###############################################################
# Plot chromosome tracks
###############################################################

for sector in circos.sectors:

    outer = sector.add_track(
        (95, 98),
        name="axis_track"
    )

    outer.axis(
        fc="none",
        ec="black",
        lw=0.5
    )

    # Major ticks every 50 Mb
    interval = 50_000_000

    outer.xticks_by_interval(
        interval=interval,
        tick_length=3,
        outer=True,
        show_label=True,
        label_size=size - 2,
        label_orientation="vertical",
        label_formatter=lambda v: f"{v / 1e6:.0f}",
        line_kws=dict(
            ec="grey"
        )
    )

    # Minor ticks every 10 Mb without labels
    outer.xticks_by_interval(
        interval=10_000_000,
        tick_length=1,
        outer=True,
        show_label=False,
        line_kws=dict(
            ec="black",
            lw=0.5
        )
    )

    # Place chromosome label at the sector center
    center_x = (sector.start + sector.end) / 2

    sector.text(
        text=sector.name.replace("chr", ""),
        x=center_x,
        r=110,
        adjust_rotation=True,
        size=size,
        color="black",
        rotation=90
    )


    ###########################################################
    # Plot segdup hotspot density (80% length subset)
    ###########################################################

    density_track = sector.add_track(
        (94, 84),
        r_pad_ratio=0.1
    )

    density_this_chr = segdup_density[
        segdup_density["Chromosome"] == sector.name
    ]

    density_track.bar(
        x=density_this_chr["Start"].to_numpy(),
        height=density_this_chr["percent_segdup"].to_numpy(),
        width=5_000_000,
        align="edge",
        vmax=100.0,
        color="gold",
        alpha=0.8
    )


    ###########################################################
    # Plot genes within segdup hotspots
    ###########################################################

    gene_hotspot_track = sector.add_track(
        (83, 73),
        r_pad_ratio=0.1
    )

    features_this_chr = gene_hotspots[
        gene_hotspots["Chromosome"] == sector.name
    ]

    features_to_plot = []

    for _, row in features_this_chr.iterrows():

        feature = SeqFeature(
            FeatureLocation(
                int(row["Start"]),
                int(row["End"])
            )
        )

        features_to_plot.append(feature)

    gene_hotspot_track.genomic_features(
        features_to_plot,
        fc="grey",
        ec="grey",
        lw=0.5,
        alpha=0.8
    )


    ###########################################################
    # Plot gene features
    ###########################################################

    gene_track = sector.add_track(
        (72, 62),
        r_pad_ratio=0.1
    )

    if sector.name in matching_features:

        for feature in matching_features[sector.name]:

            if feature.type == "gene":

                # Default color
                color = "gray"

                # Check qualifiers for genes of interest
                if hasattr(feature, "qualifiers") and feature.qualifiers:

                    for key in [
                        "gene",
                        "product",
                        "Name",
                        "ID",
                        "Parent"
                    ]:

                        if key in feature.qualifiers:

                            for value in feature.qualifiers[key]:

                                value_str = str(value)

                                for term in search_terms:

                                    if term in value_str:
                                        color = gene_colors[term]
                                        break

                                else:
                                    continue

                                break

                            else:
                                continue

                            break

                gene_track.genomic_features(
                    feature,
                    fc=color,
                    ec=color,
                    lw=0.5
                )


###############################################################
# Plot segdup links
###############################################################

bedpe = containing_links_df.copy()

# Filter for valid chromosomes
valid_chrs = list(chr_sizes.keys())

bedpe_filtered = bedpe[
    (bedpe["chr1"].isin(valid_chrs))
    & (bedpe["chr2"].isin(valid_chrs))
]


for _, row in bedpe_filtered.iterrows():

    region1 = (
        row["chr1"],
        row["start1"],
        row["end1"]
    )

    region2 = (
        row["chr2"],
        row["start2"],
        row["end2"]
    )

    contained_genes = row["contained_genes"]

    # Default link color
    color = "black"

    for gene, gene_color in gene_colors.items():

        if gene in contained_genes:
            color = gene_color
            break

    # Line width for intra- and inter-chromosomal links
    lw = 0.1 if row["chr1"] == row["chr2"] else 0.1

    circos.link(
        region1,
        region2,
        color=color,
        direction=1,
        lw=lw
    )


###############################################################
# Plot figure
###############################################################

fig = circos.plotfig()
ax = fig.axes[0]

# Widen the canvas and reserve space on the right for the legend
fig.set_size_inches(11, 9)

fig.subplots_adjust(
    left=0.04,
    right=0.75,
    bottom=0.05,
    top=0.95
)


###############################################################
# Create legend with GO subtitle inside the legend box
###############################################################

# Invisible handle for the subtitle
subtitle_handle = mpatches.Patch(
    facecolor="none",
    edgecolor="none",
    alpha=0,
    label=(
        "GO: Regulation of Protein\n"
        "Localization to Cajal body"
    )
)

combined_handles = [subtitle_handle]

for gene, color in gene_colors.items():

    combined_handles.append(
        mpatches.Patch(
            facecolor=color,
            edgecolor=color,
            label=gene
        )
    )


legend = ax.legend(
    handles=combined_handles,

    # Place the legend completely outside the Circos axes
    loc="upper left",
    bbox_to_anchor=(1.02, 1.02),

    frameon=True,
    fancybox=True,
    shadow=True,
    ncol=1,

    fontsize=10,
    title="Genes of interest",
    title_fontsize=size,

    labelspacing=0.7,
    borderpad=0.9,
    handlelength=1.5,
    handletextpad=0.6
)


# Format the main legend title
legend.get_title().set_fontweight("bold")


# Format and left-align the GO subtitle
subtitle_text = legend.get_texts()[0]

subtitle_text.set_fontsize(size - 4)
subtitle_text.set_color("#222222")
subtitle_text.set_linespacing(1.4)
subtitle_text.set_ha("left")
subtitle_text.set_multialignment("left")


###############################################################
# Track labels
###############################################################

track_labels = [
    (
        85,
        "1) Segdup density",
        "black"
    ),
    (
        70,
        "2) Gene within segdup\n"
        "    hotspots (n=1271)",
        "black"
    ),
    (
        59,
        "3) Genes of interest \n"
        "    & their paralogs",
        "black"
    ),
    (
        42,
        "4) Segdup links associated\n"
        "    w/ genes of interest\n"
        "    & their paralogs",
        "black"
    )
]


# Add labels in the gap space at 0°
for track_top, text, color in track_labels:

    ax.text(
        np.deg2rad(0),
        track_top + 2,
        text,
        rotation=0,
        ha="left",
        va="bottom",
        color=color,
        size=size - 2,
        fontweight="bold"
    )


# Do not use fig.tight_layout() here because it can move the
# legend back toward the Circos plot.

fig
```
