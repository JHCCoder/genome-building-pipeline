```python
chrom="chr2"
hor_char=pd.read_csv("/tscc/projects/ps-renlab2/jhc103/degu-genome-assembly-proj/output/outputs-from-centraAnno/hifiasm-0414/"+chrom+"_horDecomposedResult.csv",names=["chr","HOR_template","start","end","repblock_ident","HOR_size","score"],sep=",")

# Convert value counts to a DataFrame
hor_df = hor_char["HOR_size"].value_counts().rename_axis('hor_len').reset_index(name='counts')

# Get the top 10 most frequent HOR lengths
top_10 = hor_df.head(10)

# Calculate average repblock_ident for each HOR_size in the top 10
avg_identities = []
for hor_len in top_10['hor_len']:
    avg_ident = hor_char[hor_char['HOR_size'] == hor_len]['repblock_ident'].mean()
    avg_identities.append(avg_ident)

# Add average identity to the top_10 dataframe
top_10['avg_identity'] = avg_identities
```

```python
# Create a colormap based on average identity scores
cmap = plt.cm.viridis  # You can use 'plasma', 'coolwarm', 'RdYlBu' etc.
norm = plt.Normalize(top_10['avg_identity'].min(), top_10['avg_identity'].max())
colors = [cmap(norm(ident)) for ident in top_10['avg_identity']]

# Create the bar plot with your original style but colored bars
ax = top_10.plot(kind='bar', x='hor_len', y='counts', edgecolor='black', rot=0, color=colors)

# Add chart labels and title
plt.xlabel('HOR Length')
plt.ylabel('Frequency')
plt.title(chrom)

# Add colorbar to show the identity scale
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = plt.colorbar(sm, ax=ax)
cbar.set_label('Average Identity Score')

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Create a figure with subplots
fig, axes = plt.subplots(6, 5, figsize=(20, 24))
axes = axes.flatten()

# Find global min and max for identity scores across all chromosomes
all_identities = []

for chrom in chr_colors.keys():
    try:
        hor_char=pd.read_csv("/tscc/projects/ps-renlab2/jhc103/degu-genome-assembly-proj/output/outputs-from-centraAnno/hifiasm-0414/"+chrom+"_horDecomposedResult.csv",names=["chr","HOR_template","start","end","repblock_ident","HOR_size","score"],sep=",")

        hor_df = hor_char["HOR_size"].value_counts().rename_axis('hor_len').reset_index(name='counts')
        top_10 = hor_df.head(10)
        
        for mono_len in top_10['hor_len']:
            avg_ident = hor_char[hor_char['HOR_size'] == mono_len]['repblock_ident'].mean()
            all_identities.append(avg_ident)
    except:
        continue

# Create global colormap
if all_identities:
    global_min = min(all_identities)
    global_max = max(all_identities)
    cmap = plt.cm.viridis
    norm = plt.Normalize(global_min, global_max)

# Loop through chromosomes
for i, chrom in enumerate(chr_colors.keys()):
    if i >= 30:
        break
        
    try:
        hor_char=pd.read_csv("/tscc/projects/ps-renlab2/jhc103/degu-genome-assembly-proj/output/outputs-from-centraAnno/hifiasm-0414/"+chrom+"_horDecomposedResult.csv",names=["chr","HOR_template","start","end","repblock_ident","HOR_size","score"],sep=",")
        hor_df = hor_char["HOR_size"].value_counts().rename_axis('hor_len').reset_index(name='counts')
        top_10 = hor_df.head(10)
        
        avg_identities = []
        for mono_len in top_10['hor_len']:
            avg_ident = hor_char[hor_char['HOR_size'] == mono_len]['repblock_ident'].mean()
            avg_identities.append(avg_ident)
        
        top_10['avg_identity'] = avg_identities
        colors = [cmap(norm(ident)) for ident in top_10['avg_identity']]
        
        ax = axes[i]
        bars = ax.bar(range(len(top_10)), top_10['counts'], color=colors, edgecolor='black')
        
        ax.set_xticks(range(len(top_10)))
        ax.set_xticklabels(top_10['hor_len'], rotation=45,size=13)
        ax.tick_params(axis='y', labelsize=13)
        ax.set_xlabel('HOR Length',size=12)
        ax.set_ylabel('Count',size=12)
        ax.set_title(f'{chrom}',size=12)
        
        # Add individual colorbar (smaller)
        from mpl_toolkits.axes_grid1 import make_axes_locatable
        divider = make_axes_locatable(ax)
        cax = divider.append_axes("right", size="5%", pad=0.1)
        sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
        sm.set_array([])
        plt.colorbar(sm, cax=cax)
        
    except Exception as e:
        print(f"Error with {chrom}: {e}")
        axes[i].set_visible(False)

# Hide unused subplots
for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.show()
```

```python

# Initialize an empty list to store results
largest_hor_list = []

# Loop through each chromosome
for chrom in chr_colors.keys():
    try:
        # Read the HOR decomposed result file
        hor_char = pd.read_csv(
            "/tscc/projects/ps-renlab2/jhc103/degu-genome-assembly-proj/output/outputs-from-centraAnno/hifiasm-0414/" + 
            chrom + "_horDecomposedResult.csv",
            names=["chr", "HOR_template", "start", "end", "repblock_ident", "HOR_size", "score"],
            sep=","
        )
        
        # Convert value counts to a DataFrame
        hor_df = hor_char["HOR_size"].value_counts().rename_axis('hor_len').reset_index(name='counts')
        
        # Get the largest HOR (most frequent)
        if not hor_df.empty:
            largest_hor = hor_df.head(1).copy()
            largest_hor["color"] = chr_colors[chrom]
            largest_hor['chromosome'] = chrom
            largest_hor_list.append(largest_hor)
            
    except FileNotFoundError:
        print(f"Warning: File for {chrom} not found, skipping...")
    except Exception as e:
        print(f"Error processing {chrom}: {e}")

```
