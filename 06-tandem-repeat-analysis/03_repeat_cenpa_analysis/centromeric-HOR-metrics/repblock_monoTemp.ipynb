## Visualize the RepBlock-MonoTemp identity from hor decomposed result

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
hor_identity = pd.read_csv(f"{PROJ_ROOT}/output/outputs-from-centraAnno/hifiasm-0414/chr4_horDecomposedResult.csv",
                           names=["chromosomes","position","start","end","RepBlock-MonoTemp","Hor_size","score"],
                           sep=",")
# chr4,Pos:1450000:1452000:1,1450363,1450411,0.920000,49,high
# chr4,Pos:1450000:1452000:1,1450412,1450460,0.920000,49,high
# chr4,Pos:1450000:1452000:1,1450461,1450509,0.880000,49,high
# chr4,Pos:1450000:1452000:1,1450510,1450558,0.940000,49,high
# chr4,Pos:1450000:1452000:1,1450559,1450607,0.960000,49,high
# chr4,Pos:1450000:1452000:1,1450608,1450656,0.920000,49,high
# chr4,Pos:1450000:1452000:1,1450657,1450705,0.900000,49,high
```

```python
hor_identity["midpoint"]=(hor_identity["end"]+hor_identity["start"])/2
hor_identity
```

```python


# Create the scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(hor_identity['midpoint'], hor_identity['RepBlock-MonoTemp'], alpha=0.6)
plt.xlabel('Midpoint')
plt.ylabel('RepBlock-MonoTemp')
plt.title('Scatter Plot: RepBlock-MonoTemp vs Midpoint')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
start=124000000
end=134000000
df_plot = hor_identity[(hor_identity["start"]>=start) & (hor_identity["end"]<=end) ]
```

```python


# Create the scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(df_plot['midpoint'], df_plot['RepBlock-MonoTemp'], alpha=0.6)
plt.xlabel('Midpoint')
plt.ylabel('RepBlock-MonoTemp')
plt.title('Scatter Plot: RepBlock-MonoTemp vs Midpoint')
plt.grid(True, alpha=0.3)
plt.show()
```
