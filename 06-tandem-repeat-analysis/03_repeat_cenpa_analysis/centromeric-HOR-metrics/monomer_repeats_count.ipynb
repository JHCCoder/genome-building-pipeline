```python
hor_info = pd.read_csv(f"{PROJ_ROOT}/output/outputs-from-centraAnno/hifiasm-0414/cautils-chrOnly/HORs.bed",names=["seq_name","hor_name","start_pos","end_pos","num_monomers","hor_len","span_leng"],sep="\t")
# Example
# chr1    Pos:2200000:2206000:0_0 2200195 2202309 2       45      2114
# chr1    Pos:2200000:2206000:0_0 2203923 2204779 2       46      856
# chr1    Pos:2200000:2206000:0_0 2205310 2205871 2       46      561
# chr1    Pos:2200000:2206000:2_3 2200476 2200641 2       37      165
# chr1    Pos:2200000:2206000:0_2_3_4_0_0_2_3_4_2_3_4_0_0_2_3_0_0_0_0_0_2_3_0_4   2202333        2203922 25      530     1589
# chr1    Pos:3635000:3639000:4_4 3635824 3635947 2       32      123
```

```python
# Convert value counts to a DataFrame
hor_df = hor_info["hor_len"].value_counts().rename_axis('hor_len').reset_index(name='counts')

# Get the top 10 most frequent HOR lengths
top_10 = hor_df.head(10)

# Create a bar plot
top_10.plot(kind='bar', x='hor_len', y='counts', edgecolor='black', rot=0)

# Add chart labels and title
plt.xlabel('HOR Length')
plt.ylabel('Frequency')
plt.title('Top 10 HOR Lengths')
plt.show()
```

```python
# Get top 10 values directly from the Series
top_10_series = hor_info["hor_len"].value_counts().iloc[:10]

# Create the plot
top_10_series.plot(kind='bar', edgecolor='black', rot=0,figsize=(8, 5))
plt.xlabel('HOR Length')
plt.ylabel('Frequency')
plt.title('Top 10 HOR Lengths')
plt.show()
```
