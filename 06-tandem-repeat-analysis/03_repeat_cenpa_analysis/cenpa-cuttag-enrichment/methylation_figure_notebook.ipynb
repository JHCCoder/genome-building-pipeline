**Purpose:** Display the methylation figures for the CENP-A repeat-array project. This is a
*display* notebook — figures are generated by the underlying R/Python scripts and shown here
with captions (see the final cell for the generating scripts and data sources).

**Data:** WGBS, degu PFC tissue (`degu_6834_PFC_1`), CpG = CGN context, both strands
(`…CGN-both.frac.bw` / `…CGN-both.cov.bw`). Callable CpGs require coverage ≥ 5.

**Panels:**
1. **Per-bin CpG content & methylation** — merged TRF repeat arrays stratified into the
   9 period bins used by the CENP-A enrichment analysis (avg array length, CpG sites/kb,
   methylation fraction).
2. **Methylation around CENP-A cores** — per-chromosome methylation by spatial interval
   around each peak-based CENP-A core.
3. **WGBS coverage context** (supplementary) — coverage by interval; the satellite-junction
   core is sparse but very high-depth.

**Headline finding:** the **348–349 bp centromeric satellite arrays (period bin 6) are
hypomethylated (~0.48 fraction methylated vs ~0.77–0.83 in all other bins)**, consistent
with hypomethylated constitutive centromeric heterochromatin — and the CENP-A core region
similarly dips to ~0.62 (based on a sparse-but-high-depth CpG set).

## 1. Per-bin CpG content & methylation (TRF period bins)

Repeat arrays are merged TRF intervals (`bedtools merge -d 0`, never across bins) — the same
unit used for the CENP-A enrichment analysis. For each of the 9 period bins:

- **Average array length (bp)** — mean `end − start` over arrays
- **Average CpG sites / 1000 bp** — mean over arrays of reference CG density (case-insensitive
  count on the assembly, i.e. potential CpG sites)
- **CpG methylation fraction** — mean over arrays (with ≥ 1 callable CpG, `cov ≥ 5`) of the
  per-array mean WGBS fraction-methylated

Bin 6 = the 348–349 bp centromeric satellite: by far the longest arrays (mean ~142 kb) and the
**lowest methylation (~0.48)**. Note the satellite's callable-CpG fraction is low (few % of
reference CpGs are covered by WGBS), but tens of thousands of callable CpGs still support the
estimate. Full numbers below.

## 2. CpG methylation around CENP-A cores

Per-chromosome mean methylation fraction in spatial intervals around each **peak-based CENP-A
core** (short/long arm × distal 2–5 Mb / intermediate 0.5–2 Mb / proximal 0–500 kb, the core
itself, and the chromosome background). Points = per chromosome; diamonds = genome-wide pooled
mean.

Methylation is broadly high (~0.78–0.80) in the flanking regions, with a **dip in the CENP-A
core (~0.62 genome-wide pooled)** — the satellite-junction core is locally hypomethylated. The
long-arm proximal flank is also relatively low (~0.70). The core rests on very few callable
CpGs genome-wide (86), so it is shown with that caveat; the coverage panel below explains why.

## 3. Methylation of the 349-bp satellite around the CENP-A core

Mean methylation fraction of the **348-349 bp satellite** (bin-6 arrays) by spatial
interval around the peak-based CENP-A core — restricted to the satellite's own CpGs
(WGBS degu_6834 PFC, CpG both strands, cov>=5). Box = per-chromosome distribution,
points = per chromosome, diamonds = genome-wide pooled mean. The **% printed under
each region label is the 349-bp satellite occupancy** (genome-wide pooled), the same
occupancy shown for the 348-349 bp bin in `genomewide_occupancy_by_bin`.

The satellite is roughly **half-methylated (~0.48-0.51) at every distance from the
core** — the genome-wide non-satellite background is ~0.78-0.80 — so the satellite's
hypomethylation is a property of the repeat itself, not of proximity to the CENP-A
core. Occupancy is high on the short-arm flanks (28-33%) and minimal in the core
(4.7%); the core itself has only 7 callable satellite CpGs (chr4 only — that core is
100% satellite but unmappable), and the long-arm flanks rest on a single chromosome
(chr4).

## 4. WGBS coverage context (supplementary)

Mean per-CpG read coverage by interval (log10). Flanking regions show moderate, uniform
coverage; the **CENP-A core is sparse (few callable CpGs) but at very high depth (~2,700×
genome-wide pooled)** — consistent with the core sitting at the degenerate satellite-junction,
where short WGBS reads map poorly but those that do map pile up. This is why the core's
methylation estimate rests on a small, high-depth CpG set.
