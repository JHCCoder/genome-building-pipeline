```python
count_row=0
for species in animals:
    print("Processing: "+str(species))
    repeat_df=pd.read_csv("repeat_df_"+str(species)+".tsv", sep='\t')
    repeat_df["length"]=repeat_df["end"]-repeat_df["start"]
    ## Create chromosome column
    repeat_df['chromosome'] = repeat_df['sequence'].apply(
        lambda x: 'seq_additional' if str(x).lower().startswith('seq') else x
    )
    

    repeat_int= repeat_df.sort_values(by="length", ascending=False).head(top_n_sequence)
    repeat_int["index"]= repeat_int.index
    new_df = pd.DataFrame({
        'repeat_sequence': repeat_int['consensus_sequence'],
        'sequence_start_end_type': repeat_int['sequence'] + '_' + repeat_int['start'].astype(str) + '_' + repeat_int['end'].astype(str) + "_" + repeat_int['period_size'].astype(str)  + "_" + species +"_"+ repeat_int['index'].astype(str)
    })
    with open("animal_"+species+"_"+str(top_n_sequence)+"Longest_consensusSeq.fasta", "w") as f:
        for idx, row in new_df.iterrows():
            f.write(f">{row['sequence_start_end_type']}\n")
            f.write(f"{row['repeat_sequence']}\n")
```
